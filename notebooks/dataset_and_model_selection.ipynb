After the analysis of the results, we'll keep the **dataset v3** because this one contains the best number of columns about the temperatures of the different French departments. Hence, it reduces the numner of dimensions and reduces overfitting too.
More over, we'll focus on **Ridge** (resists to overfitting on new data) and **HistGradientBoostingRegressor** (faster than a Random Forest during the training)

```python
from sklearn.linear_model import Ridge
from sklearn.ensemble import HistGradientBoostingRegressor
```

```python
md.log_models(
    '../data/final_datasets/datasets_linear_models/conso_v3_linear.parquet',
    '../data/final_datasets/datasets_tree_based_models/conso_v3_tree.parquet',
    Ridge(),
    HistGradientBoostingRegressor(),
    np.arange(0, 10, 1)    
)
```

```python
mlflow.set_tracking_uri('http://localhost:5000')
experiment_name = "Best models Experiment"
experiment = mlflow.get_experiment_by_name(experiment_name)
```

```python
runs = mlflow.search_runs(
    experiment_ids = [experiment.experiment_id],
    order_by = ['metrics.root_mean_squared_error ASC'],
    max_results=None
)
```

```python
runs
```

```python
best_models = runs.sort_values(by="metrics.mean squared error")
```

#### We keep the estimator that performs the best for horizon = 9 (the furthest horizon) because it will, naturally, perform better for the closest horizons

```python
estimator1 = "Ridge"
dataset_name1 = "conso_v3_linear"
h = 9

estimator2 = "HistGradientBoostingRegressor"
dataset_name2 = "conso_v3_tree"

query1 = best_models[
    (best_models["tags.model"] == f"{estimator1}") &
    (best_models["tags.mlflow.runName"] == f"{estimator1}_{dataset_name1}_horizon_{h}")
    ]

score1 = query1["metrics.mean squared error"].iloc[0]
```

```python
score1
```

```python
query2 = best_models[
    (best_models["tags.model"] == f"{estimator2}") &
    (best_models["tags.mlflow.runName"] == f"{estimator2}_{dataset_name2}_horizon_{h}")
    ]

score2 = query2["metrics.mean squared error"].iloc[0]
```

```python
best_estimator = None
if score1 < score2 : 
    best_estimator = query1["tags.model"].unique()[0]
else : 
    best_estimator = query2["tags.model"].unique()[0]
```

```python
best_estimator
```

```python
best_runs = best_models[best_models["tags.model"] == best_estimator]
```
